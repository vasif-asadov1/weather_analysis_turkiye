## Aggregated Data

```python
# Join Fact and Dimensions using DuckDB SQL
query = """
    SELECT 
        f.*,
        d.year, d.month, d.day, d.hour,
        l.latitude, l.longitude
    FROM my_db.main.fact_climate f
    LEFT JOIN my_db.main.dim_datetime d 
        ON f.measured_at = d.measured_at
    LEFT JOIN my_db.main.dim_location l 
        ON f.city = l.city
    ORDER BY f.measured_at DESC
"""
```

```python
# Execute in DuckDB and output to a Pandas DataFrame
weather_data = con.execute(query).df()
```

```python
# Verify the combined structure
display(weather_data.head())
display(weather_data.info())
```

## Data Information

**Weather & Climate Metrics**

* **temperature_2m**: The raw air temperature measured exactly 2 meters above ground level in Celsius. There is no universal "normal" here, as the baseline fluctuates drastically depending on whether you are querying a summer day in Jakarta or a winter night in Stockholm.
* **apparent_temperature**: This is the "feels like" temperature in Celsius, which directly dictates human comfort. It calculates what the body actually experiences by combining the raw temperature with humidity and wind speed. High humidity blocks sweat from evaporating (making it feel much hotter), while high wind speeds strip away body heat (making it feel colder).
* **relative_humidity_2m**: The percentage of moisture currently held in the air. A healthy, comfortable range for humans is typically between 30% and 50%. Once humidity pushes past 70%, the air starts to feel oppressive and sticky.
* **precipitation**: The total volume of rain or snow measured in millimeters. Zero is a standard dry day. As a benchmark, anything above 10mm falling in a single hour indicates a very heavy, disruptive downpour.
* **wind_speed_10m**: The speed of the wind measured 10 meters above the ground in kilometers per hour. A reading of 10 to 20 km/h is a standard light breeze. Once this number crosses 40 km/h, you are looking at strong winds that can impact daily activities.

**Air Quality & Health Metrics**

For all pollution metrics, lower numbers are always healthier.

* **pm2_5**: Fine inhalable particles that are less than 2.5 micrometers wide. This is the most dangerous metric in your dataset. Because they are so small, these particles bypass the body's natural defenses, penetrating deep into the lungs and entering the bloodstream. The World Health Organization (WHO) states the 24-hour average must stay below **15 μg/m³**. Spikes in this column indicate severe smog, heavy industrial exhaust, or nearby wildfires.
* **pm10**: Coarse particles under 10 micrometers, such as dust, pollen, and construction dirt. The WHO safety threshold is below **45 μg/m³**. While high levels will cause immediate coughing and eye irritation, they are not as deeply toxic as PM2.5.
* **nitrogen_dioxide**: A toxic gas (NO2) measured in μg/m³ that is primarily produced by burning fossil fuels. The WHO health limit is **25 μg/m³**. Tracking this column is incredibly useful for business intelligence because NO2 levels map almost perfectly to vehicle traffic and power plant usage, allowing you to visualize exact rush-hour congestion patterns across your 15 cities.
* **european_aqi**: The European Air Quality Index. Instead of making executives read raw chemical volumes, this is a master metric that aggregates all pollutants into one standardized score. A score from **0 to 50** means the air quality is Good or Fair. Once the score crosses **100**, the air is officially Poor and hazardous to sensitive groups.

**Spatial & Temporal Keys**

* **Geography**: The `city`, `latitude`, and `longitude` columns pinpoint exactly where the physical reading occurred.
* **Time Series**: The `measured_at` timestamp is broken down into `year`, `month`, `day`, and `hour` to allow for easy seasonal grouping in Power BI.

You will notice the air quality columns only have 685,734 rows, while the weather columns have 1.66 million. This is because Open-Meteo's historical air quality archive only began recording in July 2022, whereas our  weather history goes back a full 10 years.
